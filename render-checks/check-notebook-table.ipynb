# One construct, isolated

This notebook exists to hold the single thing that JupyterLab's PDF export
cannot do: a markdown table. Everything else lives in `check-notebook.ipynb`,
which exports cleanly.

Splitting them apart means the render matrix can say two things at once —
that the notebook-to-PDF route works, and exactly which construct breaks it.
If the row below ever turns green, nbconvert has fixed the bug and
`check-notebook.ipynb` can absorb this cell again.
