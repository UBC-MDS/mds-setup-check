## 5. Mathematics -- read the source of this cell

will mostly *read* them rather than write them, but you should recognise the two forms.

On its own line, centred:

$$

**Double-click this cell.** One dollar sign either side puts maths in a sentence; two
put it on its own line. Copy the pattern when you need it.

Greek letters are written as commands: `$\alpha$` gives $\alpha$, `$\beta$` gives
$\beta$, `$\sigma^2$` gives $\sigma^2$. Write them that way rather than pasting the
character and they survive every export format.

### Question 8.2

Most of what you write in this program will be **functions**. Write one called
`block_size` that takes a sequence of block numbers and one block number, and
returns how many times that block appears. The answer is again given in the cell.

> **This question has hidden tests.** `grader.check("q2")` runs some of its cases
> in front of you and keeps the rest back until your work is marked. Look at how
> many cases it reports here, and remember that the number you see is not the
> number that count. Most MDS assignments work this way: passing the checks you
> can see is necessary, not sufficient.

````python
# Again, the answer. Replace the ... with the indented line below it:
#
#     def block_size(blocks, block):
#         return sum(1 for b in blocks if b == block)

def block_size(blocks, block):
    # BEGIN SOLUTION
    return sum(1 for b in blocks if b == block)
    # END SOLUTION

# Works on a plain list and on a column of the table alike.
block_size(courses["block"], 1)
````

````python
# Every case uses only builtins and your own answer. A test that needed something
# defined in another cell would fail for anyone who had not run that cell, and would
# blame their answer for it.
assert callable(block_size), "block_size should be a function"
assert block_size([1, 1, 2], 1) == 2, "two entries are block 1"
assert block_size([1, 1, 2], 2) == 1, "one entry is block 2"
````

````python
# HIDDEN
assert block_size([1, 1, 2], 9) == 0, "a block with no courses should give 0, not an error"
assert block_size([], 1) == 0, "an empty sequence should give 0"
````
